```python
from pathlib import Path
import json, zipfile

metrics_path = Path('model/metrics.json')
assert metrics_path.exists(), 'Training did not create model/metrics.json'
results = json.loads(metrics_path.read_text())
print('Dataset instances:', results['original_instances'])
print('Duplicates removed:', results['duplicates_removed'])
print('Train/test:', results['training_instances'], '/', results['test_instances'])
print('\nMODEL METRICS')
for model, m in results['metrics'].items():
    print(f"{model:24s} Accuracy={m['accuracy']:.4f} AUC={m['auc']:.4f} Precision={m['precision']:.4f} Recall={m['recall']:.4f} F1={m['f1']:.4f} MCC={m['mcc']:.4f}")

archive = Path('/content/dry_bean_trained_outputs.zip')
with zipfile.ZipFile(archive, 'w', zipfile.ZIP_DEFLATED) as z:
    z.write('test_data.csv')
    for f in Path('model').glob('*'):
        z.write(f)
print('\nCreated:', archive)
```
